#### Wikipedia Page Views ETL

This notebook extracts the most viewed Wikipedia pages for a given date
using the Wikimedia Pageviews API, transforms the data into JSON Lines
format, and uploads it to Amazon S3.

```python
# Import neccessary libraries

import requests
import json
import boto3
from datetime import datetime
```

```python
# Define the date for which to query data
query_date = "2025-11-18"

year, month, day = query_date.split("-")
```

```python
# Build the API to get data for the specified date
url = (
    "https://wikimedia.org/api/rest_v1/"
    f"metrics/pageviews/top/en.wikipedia/all-access/{year}/{month}/{day}"
)

url
```

```python
# Call the API to get data for the specified date
headers = {
    "User-Agent": "CEU-DataEngineering-Homework/1.0 (contact: student@ceu.edu)"
}

response = requests.get(url, headers=headers)

print("Status code:", response.status_code)

```

```python
# Parse the JSON response
data = response.json()

type(data), data.keys()
```

```python
# Extract the list of items from the data
items = data["items"]
len(items)
```
